Import Libraries

```python
import pandas as pd
```

Load Data

```python
df = pd.read_csv("CleanData.csv")
```

Sort Data

```python
df = df.sort_values(["city", "date"])
```

Create new Columns

```python
df["tomorrow_prcp"] = df.groupby("city")["prcp"].shift(-1)
```

```python
df["will_rain"] = (df["tomorrow_prcp"] > 0).astype(int)
```

```python
df["yesterday_prcp"] = df.groupby("city")["prcp"].shift(1)
```

```python
df["rain_3day"] = (
    df.groupby("city")["prcp"]
      .transform(lambda x: x.shift(1).rolling(3).sum())
)

df["rain_7day"] = (
    df.groupby("city")["prcp"]
      .transform(lambda x: x.shift(1).rolling(7).sum())
)
```

Droping Useless Columns

```python
df = df.drop(columns=['rainfall_intensity', 'wind_category', 'is_hot_day', 'is_cold_day', 'date', 'day'])
```

Saving Data to File

```python
df.to_csv('CleanData.csv', index=False)
```
