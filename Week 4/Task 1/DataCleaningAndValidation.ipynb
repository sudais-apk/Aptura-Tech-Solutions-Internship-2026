```python
df = df.drop(columns='visibility') # all values of Visibilty are missing so it is droped.
```

```python
df = df.astype({
    "date" : "datetime64[ns]",
    "season": "category",
    "city": "category",
    "region": "category",
    "rainfall_intensity": "category",
    "wind_category": "category"
})

print(df.dtypes)
```

Saving Data:

```python
df.to_csv('CleanData.csv', index=False)
```

# Observations:
1. visibilty column has all missing rows so whole column was droped.
2. no other missing values.
3. no Duplicate Values.
