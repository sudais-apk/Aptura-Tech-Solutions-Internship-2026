# EDA on Iris Data Set

Exploratory Data Analysis (EDA) is initial step in data science projects. It involves analyzing and visualizing data to understand its key characteristics and identify relationships between variables. EDA helps data scientists to discover patterns, spot anomalies and check assumptions, providing a better understanding of the dataset.

## 1. Introduction:
    Iris dataset is  multivariate data set introduced by British biologist and statistician Ronald Fisher in 1936.

### EDA Objective:

- Understand the structure dataset.
- Explore distributions and relationships between features.
- Compare the three iris species.
- Identify patterns in data.

### Data Source: 
    https://www.kaggle.com/datasets/saurabh00007/iriscsv

### 7.4 Outlier analysis

```python
Q1 = df.select_dtypes(include="number").quantile(0.25)
Q3 = df.select_dtypes(include="number").quantile(0.75)

IQR = Q3 - Q1

outliers = ((df.select_dtypes(include="number") < (Q1 - 1.5 * IQR)) | (df.select_dtypes(include="number") > (Q3 + 1.5 * IQR)))

outliers.sum()
```

## 8. Relationship Analysis

## 9. Compare feature distributions across species

### 9.1 Box Plots

```python
features = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

for feature in features:
    plt.figure(figsize=(7, 5))
    sns.boxplot(x="species", y=feature,
                 data=df,
                 palette=["red", "green", "blue"],)
    plt.title(f"{feature} Distribution by Specie")
    plt.show()
```

### 9.2 Voilon Graphs 

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x='species', y='sepal_length', data=df)
plt.title("Violin Plot for Sepal Length")
```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x='species', y='sepal_width', data=df)
plt.title("Violin Plot for Sepal Width")
```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x='species', y='petal_length', data=df)
plt.title("Violin Plot for Petal Length")
```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x='species', y='petal_width', data=df)
plt.title("Violin Plot for Petal width")
```

### 9.3 Means and Medians by Species

```python
# Median 

df.groupby("species").median(numeric_only=True)
```

# 10. Conclusion:

1. Iris setosa is easily distinguishable from the other species based on its feature values, like petal length and petal width.
2. Iris versicolor and Iris virginica have overlapping feature distributions, making them harder to distinguish, although petal measurements still provide significant discriminatory power.
3. Petal length and petal width are the most useful features for differentiating between the species.
4. sepal width is the least useful features for differentiating between the species.
5. There are no missing values.
6. There are three duplicate values.
